<b>Running inference with Flair NER models</b>

This notebook documents how to generate NER predictions aligned with the ArchOnto ontology from datasets available at: [https://doi.org/10.25747/B06H-RX96](https://doi.org/10.25747/B06H-RX96).

<b>Import the required libraries</b>

```python
from flair.data import Corpus
from flair.datasets import ColumnCorpus
from flair.models import SequenceTagger
import time
```

<b>Load ground-truth files</b>

```python
# Dataset options https://rdm.inesctec.pt/dataset/2026-008/resource/b7a1a050-ae97-4b03-8f37-7da994fb4b91
src_path="domain-gen/"

columns = {0: 'text', 1: 'ner'}
corpus: Corpus = ColumnCorpus(src_path, columns, test_file='test.tsv')
```

<b>Load Flair model</b>

```python
# Flair model options in https://rdm.inesctec.pt/dataset/2026-008/resource/c4021e16-dbd1-401b-9ef7-57afe03e4004
flair_model="flairel+word2vec.pt"
model = SequenceTagger.load(flair_model)

```

<b>Execute Flair model and save prediction file</b>

```python
dst_path="path/to/your/output/test.tsv"

start_time = time.time()
result = model.evaluate(corpus.test, gold_label_type="ner", dst_path)
end_time = time.time()
```

<b>Print execution inference time</b>

```python
print(f"Total runtime of the program is {end_time - start_time:.0f} seconds")
```
